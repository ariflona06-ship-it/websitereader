1. In fall 2025, how many UC campuses did the average applicant apply to? (round to two decimal places)


Total Applications
All individual applications across UC’s school
Made sure to filter for Fall 2025
Didn’t include the "Systemwide" row so applications weren't counted twice
Unique Applicants
Systemwide data specifically for Fall 2025 freshman applicants
Got the number of individual students who submitted an application
Math
Divided total applications by total unique applicants
Rounded the answer to two decimal places for a clean output

```python
import pandas as pd

df_discipline = pd.read_csv("uc_freshman_admission_by_discipline.csv")
df_ethnicity = pd.read_csv("uc_admissions_summary_by_ethnicity.csv")

campus_apps = df_discipline[
    (df_discipline['fall_term'] == 2025) &
    (df_discipline['broad_discipline'] == 'All disciplines') &
    (df_discipline['campus'] != 'Systemwide')
]['applicants'].sum()

total_unique_applicants = df_ethnicity[
    (df_ethnicity['fall_term'] == 2025) &
    (df_ethnicity['entrant_level'] == 'freshman') &
    (df_ethnicity['campus'] == 'Systemwide') &
    (df_ethnicity['count_type'] == 'App')
]['n'].sum()

avg_campuses = campus_apps / total_unique_applicants
print(f"Average UC campuses applied to: {avg_campuses:.2f}")
```

```python
import pandas as pd


df_disc = pd.read_csv("uc_freshman_admission_by_discipline.csv")

df_2025 = df_disc[(df_disc['fall_term'] == 2025) & (df_disc['campus'] != 'Systemwide')]

overall = df_2025[df_2025['broad_discipline'] == 'All disciplines'][['campus', 'admit_rate']].rename(columns={'admit_rate': 'overall_admit_rate'})
cs = df_2025[df_2025['broad_discipline'] == 'Computer Science'][['campus', 'admit_rate']].rename(columns={'admit_rate': 'cs_admit_rate'})

comparison = pd.merge(overall, cs, on='campus')
#comparison = pd.merge(systemwide, cs, on='campus')


comparison['admit_rate_drop'] = comparison['overall_admit_rate'] - comparison['cs_admit_rate']

comparison = comparison.sort_values(by='admit_rate_drop', ascending=False)

print(comparison.to_string(index=False))
```

5. In fall 2025, at how many of the 9 UC campuses was the White freshman admit rate HIGHER than the Hispanic/Latino(a) admit rate?  

Filter
Filtered the ethnicity summary data
Focused on Fall 2025 freshman data
Excluding systemwide
New formatted table
Reshaped the table using pivot_table
Grouped application and admission counts by campus and ethnic background
Calculate
Divided admitted students by total applicants for both groups
Got their campus admit rates
Evaluated where the White admit rate was higher
Summed up the result

```python
import pandas as pd


df_eth = pd.read_csv("uc_admissions_summary_by_ethnicity.csv")


df_2025 = df_eth[
    (df_eth['fall_term'] == 2025) &
    (df_eth['entrant_level'] == 'freshman') &
    (df_eth['campus'] != 'Systemwide')
]

pivot_df = df_2025.pivot_table(
    index='campus',
    columns=['ethnicity', 'count_type'],
    values='n',
    aggfunc='sum'
)

white_admit_rate = pivot_df[('White', 'Adm')] / pivot_df[('White', 'App')]
hispanic_admit_rate = pivot_df[('Hispanic/Latino(a)', 'Adm')] / pivot_df[('Hispanic/Latino(a)', 'App')]

comparison = pd.DataFrame({
    'white_admit_rate': white_admit_rate,
    'hispanic_admit_rate': hispanic_admit_rate,
    'white_higher': white_admit_rate > hispanic_admit_rate
})

print(comparison)
print(f"\nNumber of campuses where White admit rate was higher: {comparison['white_higher'].sum()}")
```

10. Five schools are listed on the answer form. Using UC Berkeley 2022-2025, which one most OUTPERFORMS its expected admit rate, after controlling for a-g completion, poverty, applicant GPA and school size?

Filter
Filtered dashboard dataset for UC Berkeley
Application cycles between 2022 and 2025
Targeted list of five high schools
Calculate
Grouped filtered data by high school
Averaged their admit_rate_residual
Residual = Difference between actual and expected admit rates
See which schools admitted more students
Verify / check
Sorted schools by average outperformance
Mission Senior High had the highest positive residual


```python
import pandas as pd

df_dash = pd.read_csv("dashboard_data.csv")

target_schools = [
    'HERCULES HIGH SCHOOL',
    'MISSION SENIOR HIGH SCHOOL',
    'MONTEREY TRAIL HIGH SCHOOL',
    'PHILLIP & SALA BURTON ACAD HS',
    'RANCHO SAN JUAN HIGH SCHOOL'
]

filtered_df = df_dash[
    (df_dash['campus'] == 'Berkeley') &
    (df_dash['fall_term'].between(2022, 2025)) &
    (df_dash['high_school'].str.upper().isin(target_schools))
]

outperformance = filtered_df.groupby('high_school')['admit_rate_residual'].mean().reset_index()

outperformance = outperformance.sort_values(by='admit_rate_residual', ascending=False)

print(outperformance.to_string(index=False))
```

Dashboard logic on Colab:

> 1. filter cards just from the 10-year window: 2016-2025, keep only California public high schools, and discard schools with fewer than 10 applicants, so the math doesn't mess up

> 2. compare actual rate of acceptance vs. expected rate ofacceptance

> 3. find average annual residual scores across 10 years to ensure positive and accurate results

> 4. sort the top scorers to get Cloverdale High School









```python
import pandas as pd

df_dash = pd.read_csv("dashboard_data.csv")

max_year = df_dash['fall_term'].max()
min_year = max_year - 9

public_hs_10yr = df_dash[
    (df_dash['fall_term'].between(min_year, max_year)) &
    (df_dash['school_type'].str.lower().str.contains('public', na=False)) &
    (df_dash['applicants'] >= 10)
].copy()

school_performance = public_hs_10yr.groupby(['cds_code', 'high_school', 'city', 'county']).agg(
    avg_residual=('admit_rate_residual', 'mean'),
    avg_actual_admit_rate=('admit_rate', 'mean'),
    avg_expected_admit_rate=('expected_admit_rate', 'mean'),
    avg_ag_completion=('ag_completion_rate', 'mean'),
    avg_gpa=('applicant_gpa', 'mean'),
    total_applicants=('applicants', 'sum'),
    years_counted=('fall_term', 'nunique')
).reset_index()

school_performance = school_performance[school_performance['years_counted'] >= 5]

top_overperformers = school_performance.sort_values(by='avg_residual', ascending=False)

print(f"=== TOP OVERPERFORMING CA PUBLIC HIGH SCHOOLS ({min_year}-{max_year}) ===")
print(top_overperformers[['high_school', 'city', 'avg_residual', 'avg_actual_admit_rate', 'avg_expected_admit_rate', 'years_counted']].head(10).to_string(index=False))
```
